<center><h2><b>Introdução</b></h2></center>


Este trabalho tem como obetivo desenvolver um sistema de visão por computador que seja capaz de detetar e contar veiculos em um video de uma autoestrada.

A abordagem que implementamos segue um pipeline basico de processamento de video, ele segue os seguintes passos:
 - **Estimação do Fundo** : Criamos uma imagem de estática referencia (neste caso a estrada) para  depois isolarmos os veiculos.
 - **Deteção de Pixeis Ativos** : Utilizamos tecnicas de diferença de imagem e de threshhold para segmentar os veiculos do fundo.
 - **Utilização de operadores morfológicos** : Aplicação de operadores para reduzirmos o ruido e consolidar as formas dos componentes conexos.
 - **Definição de Região de Interesse** : Restringimos a área que estamos a analizar de modo a reduzir os falsos positivos.
 - **Extração de Componentes Conexos** : Identificamos os veiculos/regiões ativas e calculamos as sua propriedades geométricas como as bounding boxes e centroides.
 - **Tracking** : Associamos temporalmente os frames consecutivos de modo a termos uma contagem consistente

# IMPORTS

```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
import math
```

# FUNÇÕES UTILITÁRIAS

De modo a modularizar o código, criámos duas funções auxiliares que nos ajudam a organizar o código em toda a sua extensão. A primeira função, 'reproduzir_frames', é responsável por reproduzir os vídeos teste com  todas as nuances que estes precisam (região de interesse, contornos dos veículos e a indexação de cada um deles). A segunda função, 'video_para_frames', centraliza o processo de conversão do vídeo em frames, ou seja, recebe um caminho de um vídeo e converte este para uma lista de frames. Esta função é util para depois usar na função anterirmente mencionada a lista de frames convertida.

Estas funções foram criadas de modo a não ter-se que repetir o mesmo código constantemente nas várias células que precisam de fazer uso delas.

```python
#Função para fazer os outputs
def reproduzir_frames(lista_frames_base, pontos_roi=None, dados_rastreamento=None, titulo="Resultado"):
    print(f"A reproduzir '{titulo}'")
    print("Pressiona a tecla 'Q' para fechar")
    
    for i, frame in enumerate(lista_frames_base):
        frame_visualizacao = frame.copy() #Fazemos copia para não esmagar o video original

        #Desenha a Regiao de Interesse (Se houver)
        if pontos_roi is not None:
            cv2.polylines(frame_visualizacao, [pontos_roi], isClosed=True, color=(0, 255, 0), thickness=2)

        #Desenha Caixas e IDs dos Carros (Se houver dados para este frame)
        if dados_rastreamento is not None and i < len(dados_rastreamento):
            veiculos_frame_atual = dados_rastreamento[i]
            
            for dados_veiculo in veiculos_frame_atual:
                #Se tiver 4 valores é só detecção se tiver 5 valores então tem o ID do veiculo
                #primeiros 4 para desenhar a caixa
                x, y, w, h = dados_veiculo[:4] 
                
                cv2.rectangle(frame_visualizacao, (x, y), (x + w, y + h), (0, 0, 255), 2)#retângulo vermelho

                # Se tivermos o 5o elemento (o ID), desenha se o texto
                if len(dados_veiculo) == 5:
   
                    cv2.putText(frame_visualizacao, f"V{dados_veiculo[4]}", (x, y - 5), 
                                cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0, 0, 0), 2)

        #Mostra o resultado
        cv2.imshow(titulo, frame_visualizacao)
   
        if cv2.waitKey(45) & 0xFF == ord('q'): #30ms entre frames
            break

    cv2.destroyAllWindows()
    print("Reprodução terminada.")

#Pega no video e transforma num array de frames - util para usar a função de cima
def video_para_frames(caminho_video):
    cap = cv2.VideoCapture(caminho_video)
    frames = []
    
    if not cap.isOpened():
        print("ERRO - Não foi possível abrir o vídeo!")
        return []

    while True:
        bReturn, frame = cap.read()
        if not bReturn:
            break
        frames.append(frame)
    
    cap.release()
    print(f"Vídeo carregado - {len(frames)} frames")
    return frames
```

<center><h2><b>Estimação de imagem de fundo</b></h2></center>


A estimação de uma imagem de fundo limpa é muito importante para sistemas como este de videovigilancia com camara fixa. Para detetar objetos em movimento em um situação em que o fundo é constante a estratégia mais usada é a de Subtração do Fundo. Mas para fazermos isso é necessário termos um modelo de referencia do cenário vazio, que neste caso é só a estrada. 

Tendo em conta que não nos é fornecido uma imgem da estrada sem nenhum carro, então temos que estimar a imagem temporalmente. Para isso usamos uma filtragem temporal com filtro de mediana.

Se calcularmos a mediana de cada fixel ao longo de todos os frames, então conseguimos remover os objetos que passam rapidamente. Isto porque em média cada pixel passa mais tempo a mostrar "o fundo" do que a mostrar o carros.

```python
video_original_frames = video_para_frames("AutoEstrada.avi")#converte o video para uma lista de frames

def estimar_fundo(lista_frames):
    if not lista_frames:
        print("Erro: A lista de frames está vazia.")
        return None

    print(f"A calcular a mediana de {len(lista_frames)} frames...")

    # Calcula a mediana ao longo do tempo (no axis=0)
    # A lista de frames é convertida logo para um array numpy aqui
    fundo = np.median(lista_frames, axis=0).astype(dtype=np.uint8)

    return fundo

imagem_fundo = estimar_fundo(video_original_frames)

# 3. VISUALIZAÇÃO
if imagem_fundo is not None:
    plt.figure(figsize=(10, 6))
    plt.imshow(cv2.cvtColor(imagem_fundo, cv2.COLOR_BGR2RGB))
    plt.title("Imagem de Fundo (Calculada em Memória)")
    plt.axis('off')
    plt.show()
```

Na estimação de imagem de fundo, decidimos percorrer os frames todos sendo que o vídeo em análise não é muito grande, mas para vídeos de maior dimensão teríamos de optar por outra estratégia de modo a não compremeter o custo computacional da máquina que correrá o código. Opções possíveis seriam:
Percorrer os frames do vídeo de passo em passo (ex: 2 em 2, 5 em 5, etc) ou selecionar aleatoriamente um conjunto de frames do vídeo.

**Resultado Obtido** : A imagem que resulta apresenta só a autoestrada sem qualquer veiculo. Este é o resultado que estavamos à espera porque em qualquer posição (x,y) da imagem, a presença do carro é temporalmente curto quando comparamos com o tempo em que se vê a autoestrada. Podemos concluir que o filtro de mediana elimina bem os elementos atipicos temporalmente.

<center><h2><b>Deteção de Pixeis Ativos</b></h2></center>

<center>
<img src="Formula da diferença absoluta.png"
</center>



Para realizar a deteção de pixeis ativos, decidimos basear-nos na diferença absoluta entre o frame atual e a imagem de fundo  presente nos slides (Capítulo 7) lecionados pelo professor. Primeiramente convertemos a imagem de fundo anteriormente estimada para a escala de cinzentos bem como todos os frames do vídeo. De seguida, convertemos tanto a imagem de fundo como os frames para float de modo a permitir números negativos corretamente na subtração. Posteriormente, aplicamos a fórmula da diferença absoluta. Por fim, aplicamos o limiar para determinar quais os pixeis que são considerados ativos (valores acima do limiar) e quais os que não são (valores abaixo do limiar).

Quanto ao valor do limiar que decidimos utlizar, é de boa prática não usar um valor extraído por tentativa e erro, mas sim um valor calculado com base nos dados do vídeo. Para isso, baseamo-nos na estatística da imagem de fundo, mais especificamente no desvio padrão dos valores dos pixeis da imagem de fundo. Assim teremos uma noção da complexidade ou contraste natural da autoestrada presente no video. Usaremos 50% do desvio padrão como limiar pois o desvio padrão total inclui o alto contraste das linhas brnacas presentes no video, o ruído que queremos filtrar é tipicamente inferior a metade dessa variação global. Portanto, o limiar a utilizar é adaptativo em relação ao conteúdo do vídeo, o que é mais robusto do que um valor fixo arbitrário.

Este limiar é definido dinamicamente dentro da função permitindo assim que o limiar se adapte e seja diferente para diferentes vídeos.


```python
def detetar_pixeis_ativos(lista_frames, imagem_fundo):

    if not lista_frames:
        print("Erro: A lista de frames está vazia.")
        return []

    #Garantimos que o fundo está em grayscale
    if len(imagem_fundo.shape) == 3:
        fundo_gray = cv2.cvtColor(imagem_fundo, cv2.COLOR_BGR2GRAY)
    else:
        fundo_gray = imagem_fundo
    
    #Convertemos para float - para depois usar o np.abs() - se deixarmos em uint8, numeros negativos vão dar a volta 
    fundo_float = fundo_gray.astype(float)

    ##CALC DO LIMIAR
    sigma = np.std(fundo_float) #Desvio padrão global
    limiar = sigma * 0.5 #Usamos 50% do desvio padrão por causa do alto contraste do branco das linhas com o preto da estrada que ia aumentar muito o limiar
    
    print(f"Debug:")
    print(f"Limiar Calculado: {limiar:.2f}")
    print(f"---------------------")

    lista_frames_processados = []

    #percorre os frames um a um
    for frame in lista_frames:
        #Passa frame para grayscale e float
        frame_cinzento = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
        frame_float = frame_cinzento.astype(float)
        
        #Faz a diferença absoluta |Fundo - Frame|
        diferenca_abs = np.abs(fundo_float - frame_float)
        
        # Aplica Limiar
        pixeis_ativos = np.zeros_like(diferenca_abs, dtype=np.uint8)#array de zeros e voltamos a uint8
        pixeis_ativos[diferenca_abs > limiar] = 255 #Se a diferença for maior q o limiar então passa para branco
        
        lista_frames_processados.append(pixeis_ativos)#guarda
    
    #Retorna os frames com 0 os pixeis inativos e 1 os pixeis ativos
    return lista_frames_processados

#_______OUTPUT-TESTE___________
frames_pixeis_ativos = detetar_pixeis_ativos(video_original_frames, imagem_fundo)
print("PIXEIS ATIVOS")
reproduzir_frames(frames_pixeis_ativos, titulo="Resultado - Pixeis Ativos")
```

**Resultado Obtido** : O output desta fase é uma sequencia de imagens binárias onde os veiculos aparecem como aglomerados de pixeis brancos e o fundo a preto. No entanto temos algum problemas, nomeadamente a presença de vários pontinhos brancos dispersos. Podemos assumir que este ruido deve acontecer devido a pequenas variações de iluminação, vibrações da câmara ou movimentos de elementos que são "estáticos" como as arvores ao fundo.

Estes artefactos tem que ser filtrados posteriormente para evitarmos que tambem sejam classificados como veiculos.

<center><h2><b>Operadores Morfológicos</b></h2></center>

Após a deteção de pixeis ativos, podemos observar algum ruído no conjunto de imagens binárias obtidas (vídeo) como por exemplo, pequenos pontos que correspondem às linhas a trajetado das vias onde os carros circulam e de algumas árvores presentes no vídeo que ao apresentarem pouco mas algum movimento no vídeo são detetadas também como pixeis ativos. Para mitigar este ruido de forma a que no futuro não seja classificado , e apenas sim os veículos, precisamos de aplicar alguns operadores morfológicos.

```python
def aplicar_operadores_morfologicos(video_entrada):
    #Recebe uma lista de frames binários correspondente aos pixeis ativos.

    frames_operadores_morfologicos = [] #lista onde se vai guardar os frames dos pixeis ativos depois de aplicar os operadores

    #ELEMENTOS ESTRUTURANTES/KERNELS
    #Circulo pequeno para eleminar os pontinhos de ruido em interferir muito com os carros
    kernel_ruido = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (3, 3))
    
    #Retangulo um pouco maior para deixar os carros mais cheios assim não corremos o risco dos carros ficarem cortados ao meio
    #RECT porque os carros são retangulares
    kernel_uniao = cv2.getStructuringElement(cv2.MORPH_RECT, (8, 8))

    for frame in video_entrada:

        # OPEN - Remove os pontinhos brancos isolados de ruido
        frame_sem_ruido = cv2.morphologyEx(frame, cv2.MORPH_OPEN, kernel_ruido)

        #CLOSE - Preenche buracos dentro dos objetos brancos (carros)
        frame_solido = cv2.morphologyEx(frame_sem_ruido, cv2.MORPH_CLOSE, kernel_uniao)

        frames_operadores_morfologicos.append(frame_solido)

    return frames_operadores_morfologicos

#_______OUTPUT-TESTE___________
video_morfologia = aplicar_operadores_morfologicos(frames_pixeis_ativos)
print("OPERADORES MORFOLÓGICOS APLICADOS")
reproduzir_frames(video_morfologia, titulo="Resultado - Operadores")
```

A estratégia que nós adotamos envolveu o uso de dois passos de aplicação de operadores morfologicos diferentes com elementos estruturantes diferentes:

**OPEN** :
- A operação consiste em uma erosão seguida de uma dilatação
- Utilizamos como elemento estruturante um circulo pequeno (Elipse 3x3) - Mais adequado para os pontinhos pequenos
- Como objetivo desta aplicação queremos remover os pixeis isolados a branco (pontinhos) de ruido que foram gerados no passo anterior com a erosão e a dilatação vai servir para recuperar a forma como os carros foram afetados por essa erosão. Abaixo temos uma imagem que exemplefica a utilização deste operador.

  ![OPENexample.png](OPENexample.png)

**CLOSE** :
- A operação consiste em uma dilatação seguida de uma Erosão
- Utilizamos como elemento estruturante um quadrado maior 8x8 (Rect 8x8) - Adequado tendo em conta o formato rectangular dos veiculos.
- Como objetivo desta aplicação queremos tentar preencher os buracos que estão dentro das regiões dos veiculos e também tentar juntar os carros que se fragmentaram para depois ser mais facil de fazer a sua detecção. Abaixo temos uma imagem que exemplefica a utilização deste operador.

  ![CLOSEexample.png](CLOSEexample.png)
  

**Resultado Obtido** : Após aplicarmos os operadores necessários, podemos ver que no vídeo resultante, o ruído foi reduzido significativamente, mantendo apenas assim os pixeis ativos que queremos classificar (veículos). Podemos visualizar regiões brancas e compactas que correspondem aos veiculos. Um efeito secundário que obtivemos e que é bastante notavel é que devido à perspetiva, os veiculos muito distantes e proximos uns dos outros acabam por se fundir em uma unica região conexa na operação de CLOSE. No entanto tendo em conta o proximo passo da definição da Região de Interesse em que os carros estão mais proximos da câmara e separados, esse efeito acaba por ser irrelevante.

<center><h2><b>Deteção de Regiões Ativas</b></h2></center>

A nossa ideia foi criar uma região de interesse (ROI) que abranja todas as vias da autoestrada, de modo a que apenas os veículos que passem por essa região sejam posteriormente contabilizados. Definimos os valores de cada um dos 4 pontos do trapézio "à mão" com base nas proporções da estrada no vídeo. Tentamos nos aproximar o maximo possivel dos exemplos visuais presentes no enunciado. Este passo é importante para evitar processar veiculos que passem na faxa de rodagem oposta, ruido que tenha sobrado e veiculos distantes que estão menos segmentados.

```python
#Defenição dos 4 cantos do polígono onde vamos detetar os carros
#(0,0) - canto superior esquerdo

#Canto Inferior Esquerdo
p1 = [25, 220] 

#Canto Superior Esquerdo
p2 = [98, 80]

#Canto Superior Direito
p3 = [160, 80]

#Canto Inferior Direito
p4 = [215, 220]

#array com os pontos da nossa região de interesse
pontos_roi = np.array([p1, p2, p3, p4], np.int32)

#_______OUTPUT-TESTE___________
reproduzir_frames(video_original_frames, pontos_roi, titulo="Teste da Regiao de Interesse")
```

Podemos observar um output com o vídeo orginal com a região de interesse desenhada a verde.

Com a região de interesse definida corretamente, podemos agora avançar para a fase de detetar as regiões ativas (veículos) que passam por essa região criada anteriormente.

```python
def extrair_componentes_conexos(video_binario, pontos_roi):
    
    lista_bounding_boxes = []

    for frame_bin in video_binario:
        #lista das bounding boxes / caixas que envolvem os carros em 1 frame
        caixas_neste_frame = []
        
        #EXTRAIMOS OS COMPONENTES CONEXOS 
        #RETR_EXTERNAL - Ignora buracos dentro dos carros que temos ainda depois de aplicar os operadores 
        #CHAIN_APPROX_SIMPLE - guarda só os pontos essenciais
        contornos, _ = cv2.findContours(frame_bin, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
        
        for contorno in contornos:
            x, y, w, h = cv2.boundingRect(contorno)#pega no rectangulo q envole o contorno

            #Calcula o centro do rectangulo
            centro_x = int(x + w / 2)
            centro_y = int(y + h / 2)

            #pointPolygonTest devolve se um ponto está dentro ou fora de um poligno
            #valor positivo - está dentro ; valor negativo - está fora ; 0 - está na borda
            #o parametro false diz à função que não precisa de calcular a distancia
            if cv2.pointPolygonTest(pontos_roi, (centro_x, centro_y), False) >= 0:
                #Se estiver dentro então guardamos 
                caixas_neste_frame.append((x, y, w, h))
        
        #Guardamos a lista de carros de cada frame na lista de frames
        lista_bounding_boxes.append(caixas_neste_frame)

    return lista_bounding_boxes

#_______OUTPUT-TESTE___________
dados_veiculos = extrair_componentes_conexos(video_morfologia, pontos_roi)

print("Extração dos Componentes Conexos")
#VISUALIZAR COM O VIDEO ORIGINAL
reproduzir_frames(video_original_frames, pontos_roi, dados_veiculos, titulo = "Resultado da Extracao dos Componentes")

#VISUALIZAR COM O VIDEO DOS OPERADORES MORFOLOGICOS
#video_morfologia_rgb = [cv2.cvtColor(frame, cv2.COLOR_GRAY2BGR) for frame in video_morfologia] #passa pa rgb para poder adicionar os contornos coloridos novos
#reproduzir_frames(video_morfologia_rgb, pontos_roi, dados_veiculos, titulo = "Resultado da Extracao dos Componentes")
```

Nesta fase fizemos a etiquetagem de componentes conectos em que o objetivo é agrupar os pixeis ativos vizinhos em objetos distintos.

Para extrairmos os componentes conexos dos pixeis ativos, decidimos utilizar a função **cv2.findContours** da biblioteca OpenCV. Esta função é eficiente e amplamente utilizada para detetar contornos de componentes conexos em imagens binárias. 

Como parametro do findContours utilizamos:
 - **cv2.RETR_EXTERNAL**: Este parametro é importante na nossa implementação pois ele recupera os contornos externos e ignora os buracos dentro dos veiculos, buracos estes que são possiveis observar em alguns momentos depois da aplicaçãos dos operadores.
 - **cv2.CHAIN_APPROX_SIMPLE**: Este é um parametro que é obrigatório preencher do findContours, esta opção faz com que apenas seja guardado pontos essenciais, desta forma optimiza o uso de memória.

Ao aplicar esta função às imagens processadas (após a aplicação dos operadores morfológicos), conseguimos identificar as diferentes regiões ativas que correspondem aos veículos presentes na região de interesse.

Para cada contorno que detetamos, calculamos a sua bounding box / caixa envolvente e o seu respectivo centroide(ponto central da caixa).

Usamos os centroides das bounding boxes calculadas para representar a posição dos veículos em cada frame de modo a poder observar os contornos ao longo de toda a extensão da região de interesse. Recorremos à função **cv2.pointPolygonTest** para verificar se o centroide de cada caixa/boundingbox se encontra dentro da região de interesse. Se o centroide estiver dentro da região ou na sua borda. Se esta condição verificar-se desenhamos o contorno no frame atual. Esta verificação é feita para cada frame do vídeo. Se não usassemos esta verificação, outros veiculos da faixa de rodagem oposta e tambem as junções de veiculos (quando está muito longe da camara) verificadas no passo anterior seriam detetadas ou até mesmo pontos de ruído que não tenham sido totalmente elinados pelos operadores morfológicos aplicados anteriormente.
Quando os veículos se ausentam da região de interesse, os contornos deixam de ser desenhados. Todos os veículos que circulem fora da regiáo de interesse são totalmente ignorados para efeitos de contagem. 

**Resultado Obtido** : Podemos ver que as Bounding Boxes desenhadas acompanham corretamente os veiculos  apenas dentro dos limites da região de interesse. Para efeitos de teste e nitidez do ambiente envolvente, usamos o video original como base para desenhar os contornos dos veículos detetados, como fizemos anteriormente com a região de interesse. 

No entanto verificamos que por breves momentos quando os veiculos estão a sair da nossa região de interesse pode acontecer uma intermitência em que visualmente as caixas "piscam". Isto acontece porque o centroide oscila momentaneamente para dentro e para fora da ROI, seja por causa do redimensionamento da caixa ou do seu posicionamento, isto faz com que a caixa desaparça e reapareça em frames seguidos. Logo este é um problema que temos que resolver no passo seguinte para termos um bom rastreamento dos veiculos.

De seguida passamos à fase da contagem dos veículos.

<center><h2><b>Correspondência de regiões entre imagens consecutivas</b></h2></center>

Após termos realizado a deteção de regiões ativas, o próximo passo é estabelecer a correspondência entre essas regiões em imagens consecutivas. Para isso, decidimos que usar um método baseado na distância entre as boundingboxes das regiões ativas em frames consecutivos seria uma boa ideia para conseguir distinguir cada veículo. Isto porque a contagem de veiculos temos que saber que o objeto A no frame t é o mesmo objeto A no frame t+1.

```python
#Constantes de controlo da função

MAX_DIST_ENTRE_BBOXES = 25
FRAMES_PRESISTENCIA = 30

def correspondencia_imgs_consecutivas(lista_bboxes_por_frame):   
    resultado_rastreamento = []
    proximo_id_disponivel = 1

    carros_ativos = {}#Veiculos detectados no frame anterior   
    carros_perdidos = {}#Veiculos que desapareceram do frame á menos de "FRAMES_PRESISTENCIA"   

    for idx, caixas_no_frame in enumerate(lista_bboxes_por_frame):     

        #PREPARAÇÃO
        veiculos_atuais = [] #veiculos detactados no frame que se está a percorrer
        for (x, y, w, h) in caixas_no_frame:
            veiculos_atuais.append({
                'box': (x, y, w, h), 
                'position': (x, y), #Assumimos o canto superior esquerdo como a posição do carro
                'id': None
            })

        #FRAME 0
        #Inicialmente detectamos todos os veiculos presentes no frame inicial
        #e ordenamos de modo a que os que estão mais longe da camara têm um indice mais baixo
        if idx == 0:
            lista_para_ordenar = [] #array temporario para ordenar os carros no primeiro frame

            for veiculo in veiculos_atuais:
                # Usamos o Y da posição
                y_topo = veiculo['position'][1]
                lista_para_ordenar.append((y_topo, veiculo))
            
            lista_para_ordenar.sort()#ordena pela altura

            # Redefinimos o array de veiculos_atuais já ordenado
            veiculos_atuais = []

            for y, veiculo in lista_para_ordenar:
                veiculos_atuais.append(veiculo)
                
            # Atribuimos os primeiros IDs
            for veiculo in veiculos_atuais:
                veiculo['id'] = proximo_id_disponivel
                proximo_id_disponivel += 1

        # FRAMES SEGUINTES - Distância e Persistência
        else:
            # Primeiro identificamos os carros tendo em conta a sua distancia relativamente ao frame anterior
            ids_ja_associados = []
            detecoes_ja_associadas = [] 
            
            #percorremos os carros anteriores
            for id_antigo, dados_antigos in carros_ativos.items():
                
                melhor_dist = MAX_DIST_ENTRE_BBOXES
                melhor_indice = -1 #inicialmente nenhum
                
                pos_antiga = np.array(dados_antigos['position'])#convetemos para np.array para usar a formula da dist eu clideana
                #comparamos a distancia dos carros do frame anterior aos do frame atual
                for i, veiculo in enumerate(veiculos_atuais):
                    
                    if i in detecoes_ja_associadas:
                        continue         

                    pos_atual = np.array(veiculo['position'])
                    dist = math.dist(pos_atual, pos_antiga)
                    if dist < melhor_dist:
                        melhor_dist = dist #se for menor então atualizamos
                        melhor_indice = i

                if melhor_indice != -1:
                    veiculos_atuais[melhor_indice]['id'] = id_antigo
                    ids_ja_associados.append(id_antigo)
                    detecoes_ja_associadas.append(melhor_indice)


            # Depois tentamos corresponder se um carro que "apareceu" está muito proximo de um que acabou de desaparecer
            #Se sim, significa que é o mesmo carro e brevemente ele deixou de ser detectado
            for i, veiculo in enumerate(veiculos_atuais):

                if veiculo['id'] is None: 
                    melhor_dist_perdido = MAX_DIST_ENTRE_BBOXES
                    id_recuperado = None
                    pos_atual = np.array(veiculo['position'])

                    for id_perdido, dados_perdidos in carros_perdidos.items():
                        
                        pos_perdida = np.array(dados_perdidos['position'])
                        dist = math.dist(pos_atual, pos_perdida)
                        if dist < melhor_dist_perdido:
                            melhor_dist_perdido = dist
                            id_recuperado = id_perdido        

                    if id_recuperado is not None:
                        
                        veiculo['id'] = id_recuperado
                        
                        # Como não podemos apagar items do dicionário enquanto iteramos, 
                        # fazemos o del logo a seguir (aqui é seguro pois temos o break implícito pelo fluxo ou controle posterior)
                        del carros_perdidos[id_recuperado]

            #Cria se IDs para quem é realmente novo
            for veiculo in veiculos_atuais:
                if veiculo['id'] is None:
                    veiculo['id'] = proximo_id_disponivel
                    proximo_id_disponivel += 1

        # ATUALIZAÇÃO DE ESTADO       
        novos_carros_ativos = {}
        output_frame = [] 

        # Guardar o estado atual dos detetados
        for veiculo in veiculos_atuais:
            id_carro = veiculo['id']
            box = veiculo['box']
            posicao = veiculo['position']

            novos_carros_ativos[id_carro] = {'position': posicao, 'box': box}
            output_frame.append((box[0], box[1], box[2], box[3], id_carro))

        #Gere quem desaparece (Frames > 0)
        if idx > 0:
            
            for id_antigo, dados_antigos in carros_ativos.items():
                if id_antigo not in novos_carros_ativos:
                    # Move para a lista de perdidos
                    carros_perdidos[id_antigo] = {
                        'position': dados_antigos['position'], 
                        'ttl': FRAMES_PRESISTENCIA
                    }
                    
        #Decrementa o TTL e remove os que passaram
        ids_para_remover = []
        for id_perdido in carros_perdidos:
            carros_perdidos[id_perdido]['ttl'] -= 1
            if carros_perdidos[id_perdido]['ttl'] <= 0:
                ids_para_remover.append(id_perdido)    

        for id_rem in ids_para_remover:
            del carros_perdidos[id_rem]

        # Atualiza o dicionário principal
        carros_ativos = novos_carros_ativos
        resultado_rastreamento.append(output_frame)

    return resultado_rastreamento
#OUTPUT-TESTE__________

print("Correspondencia de Regiões entre imagens consecutivas")
dados_com_veiculos_identificados = correspondencia_imgs_consecutivas(dados_veiculos)
reproduzir_frames(video_original_frames, pontos_roi, dados_com_veiculos_identificados, titulo="Resultado Final")
```

O nosso algoritmo implementado neste paso é baseado na distancia espacial e memoria temporal:

 - Correspondencia de veiculos através da Distancia Euclidiana:
   
Para isto partimos do pressuposto que quando a taxa de frames é alta que chegue, o deslocamento de um veiculo entre frames é pequeno.

Calculamos a distancia Euclideana entre a posição de um veiculo (canto superior esquerdo da boundingbox - utilizamos esta posição apenas por conveniencia, não havia necessidade de voltar a calcular o centroide) no frame atual com a posição dos veiculos registrados no frame anterior. Usamos a distancia Euclideana porque esta é uma métrica comummente usada para medir a distância entre dois pontos num espaço bidimensional, calculamos através da função **math.dist** que já tinhamos prática na sua utilização.

Se a distancia calculada for inferior a **MAX_DIST_ENTRE_BBOXES** (constante esta que foi definida como 25 pixeis pois esta é uma distancia curta que chegue para não assumir um carro que esteja na via do lado ou atrás/frente como o mesmo) então assumimos que é o mesmo veiculo e mantemos o seu ID.

- Gestão das falhas de persistência:

Para resolvermos o problema falado no passo anterior de intermitencia quando os veiculos saem da ROI, decidimos implementar um sistema de  time to live.

Em suma, quando um veiculo com ID já atribuido deixa de ser detetado, não o eliminamos logo, em vez disso movemos o para uma lista de veiculos que acabaram de "desaparecer". 
Se um veiculo novo for detetado dentro de um espaço de 30 frames (**FRAMES_PRESISTENCIA** - tempo curto de time to live que chegue para não assumir o proximo carro a sair na mesma via como um que acabou de sair) depois de um veiculo "desaparecer" e estiver muito perto da ultima zona em que esse carro desaparecido saiu da ROI então sabemos que é o mesmo. Isto funciona porque os novos carro que são detetados estão muito mais perto da cârama do que os carros que acabaram de sair.


- Ordenação Inicial
  
Ao longo do desenvolvimento da função notamos que inicialmente, o carro que está à frente estava a ser identificado como V4 e os que estavam atrás como V1,V2 e V3. Para corrigir isto, tivemos a necessidade de no primeiro frame, ordenadar os carros de forma correta pois o resultado que queremos, e que faz mais sentido, é que o carro mais à frente na zona da região de interesse tenha sido o primeiro a entrar na mesma portanto queremos que este seja o veículo 1 (V1) e depois assim sucessivamente para os restantes carros presentes no frame 0. Esta ordenação é feita com base na coordenada y de cada veículo, ordenando do menor para o maior valor de y (de cima para baixo na imagem).

**Resultado Final** : O algoritmo que desenvolvemos podemos observar que consegue seguir os veiculos ao longo da nossa Região de Interesse e mantem os IDs de cada veiculo estáveis o que permite uma boa contagem dos veiculos no fluxo de tráfego.

# OBJETIVO OPCIONAL - Estimar a velocidade dos veículos e a velocidade média em cada via da autoestrada

<center><h2><b>Contextualização do Objetivo Opcional</b></h2></center>

Um dos desafios em sistemas de visão  aplicados ao tráfego é a extração de métricas como velocidade e distância a partir de imagens 2D. 

O problema é a Projeção de Perspetiva, onde a escala da imagem varia com a profundidade, ou seja, um objeto distante ocupa menos pixéis do que um objeto próximo, mesmo que percorra a mesma distância. Para resolver este problema e calcular a velocidade real em km/h dos veículos, e também a velocidade média por via, implementámos uma transformação geométrica (Homografia) de modo a obter uma vista de topo. 

Com base em informações do IMT sabemos que a largura de cada via numa autoestrasa é de 3,75 metros, como podemos ver na imagem abaixo:

![CaracteristicasAutoestrada.jpg](CaracteristicasAutoestrada.jpg)


Com base em informações retiradas do Manual de Sinalização e Planeamento das Acessibilidades e da Gestão Viária sabemos também que o tamanho de um traço descontinuo é de 4 metros e com 10 metros de separação entre eles.

![InformacoesLinhasLongitudinais.jpg](InformacoesLinhasLongitudinais.jpg)


# Pressupostos assumidos

Assumimos que a câmara está alinhada com o eixo da estrada. Desta forma, a distorção da perspetiva é basicamente apenas longitudinal, em que por estarmos a ver a estrada de frente, a sua altura é bem menor do que se a estivessemos a ver por cima. Por esta razão, mantemos a largura da imagem transformada igual à largura da região de interesse na imagem original. Se a câmara estivesse rodada mais para a esquerda ou para a direita, teríamos que ajustar a largura da imagem transformada para refletir a distorção lateral. Abaixo tem uma imagem que gerámos com uma ferramente de IA (Nano Banana) para exemplificar esta distorção na altura que temos:

<img src="EsquemaVisaoEstrada.png" alt="Esquema de Distorção de Perspetiva" width="800">

Segundo a célula anterior, sabemos que o tamanho real da largura de uma via da autoestrada é de 3.75 metros, dos traços longitudinais descontínuos 4 metros e do espaço entre eles 10 metros. Assim, ao contarmos o número de espaços entre os traços na região de interesse, podemos determinar, no plano visto de cima, quantas vezes maior é a altura da região de interesse comparada com a largura das 3 vias. Com isto, conseguimos estabelecer uma relação de escala entre os pixeis na imagem e as distâncias reais em metros, mantendo assim uma boa proporção. 

<center><h2><b>Calibração e Parâmetros Reais</b></h2></center>

De seguida, precisamos de calcular os parâmetros reais do sistema de projeção de modo a conseguirmos fazer a transformação de perspetiva corretamente. Para isso, usamos como **src_points** os 4 pontos que definem a nossa região de interesse no vídeo original.

Manuseamos estes pontos para calcular a largura totalem pixeis da região de interesse na imagem original (**LARGURA_PERSPETIVA**) pois esta será igual à largura da imagem transformada como já referido anteriormente.

Sabemos que a largura real em metros da região de interesse é igual a 3 vezes (3 vias) a largura de uma via (3.75 metros). A altura real em metros é calculada com base no número de traços e no número de espaços existentes entre eles na região de interesse. Observando visualmente a região de interesse, podemos contar que existem aproximadamente 3 traços e 3.5 espaços entre eles.

Assim, sabendo a altura e largura real em metros, podemos calcular a porporção que poderemos aplicar à imagem transformada de modo a evitar distorções dos veículos na estrada.

Sabendo a **LARGURA_PERSPETIVA** e a **ALTURA_PERSPETIVA**, podemos definir os **dst_points** que serão os pontos que definem o retângulo que representará a vista de topo das imagens transformadas.

Sem este tratamento de dados, caso atribuíssemos valores arbitrários para a largura e altura das imagens transformadas, a escala entre os eixos X e Y não seria a mais consistente, o que levaria a medições não tão precisas de distância e velocidade.


```python
#Pontos de Origem
src_points = np.float32([p2, p3, p1, p4]) #Superior Esquerdo, Superior Direito, Inferior Esquerdo, Inferior Direito
#print(src_points)

# Definimos a largura e altura da imagem em perspectiva
LARGURA_PERSPETIVA = (p4[0]-p1[0]) #Canto inferior esquerdo - canto inferior direito

n_vias = 3
#largura_real = 3,75m * 3vias = 11,27metros
largura_real = 3.75 * n_vias #metros

tamanho_linha = 4
espaco_entre_linha =10

#Ao observar a imagem vemos que temos 3 linhas e 3 espaços e meio dentro da roi longitudinalmente
altura_real = (tamanho_linha * 3) + (espaco_entre_linha * 3.5)

proporcao_altura_largura = altura_real/largura_real
#print(proporcao_altura_largura)

#round para ficar inteiro
ALTURA_PRESPETIVA = round(LARGURA_PERSPETIVA * proporcao_altura_largura)


# Pontos de Destino (formam um retangulo em vez de um trapézio)
#Estas funções do open cv trabalham com valores em float32 para materem a precisão
dst_points = np.float32([
    [0, 0],                  #Superior Esq
    [LARGURA_PERSPETIVA, 0],     #Superior Dir
    [0, ALTURA_PRESPETIVA],      #Inferior Esq
    [LARGURA_PERSPETIVA, ALTURA_PRESPETIVA] #Inferior Dir
])
```

<center><h2><b>Transformação de Perspetiva (Homografia)</b></h2></center>

Com os pontos de origem (**src_points**) e destino (**dst_points**) definidos, podemos calcular a **matriz_homografia** , para isso exploramos a função `cv2.getPerspectiveTransform` da biblioteca OpenCV. 

Com a matriz de homografia calculada, aplicamos dentro da função **aplicar_perspetiva_video** a transformação H a cada frame do vídeo original, que percorrido dentro de um ciclo for, aplica esta matriz a todos os frames do vídeo original com o uso de outra função do OpenCV, a`cv2.warpPerspective`.

Esta aplicação da perspetiva é feita segundo a matriz de homografia calculada, sobre o vídeo original com determinado tamanho de saída (**LARGURA_PERSPETIVA** x **ALTURA_PERSPETIVA**)

É de notar que a transformação só é válida para o plano mesmo da estrada em que Z=0. Objetos com altura (como os carros) sofrem uma distorção, em que parecem "deitados" ou achatados.

```python
# Calcula a Matriz de Homografia H com func do cv2
matriz_homografia = cv2.getPerspectiveTransform(src_points, dst_points)

def aplicar_perspetiva_video(lista_frames, matriz, tamanho_saida):
    frames_transformados = []
    
    for frame in lista_frames:
        #Aplica a transformação H à imagem inteira
        frame_warped = cv2.warpPerspective(frame, matriz, tamanho_saida)
        frames_transformados.append(frame_warped)
        
    return frames_transformados

# _______OUTPUT-TESTE___________
video_vista_topo = aplicar_perspetiva_video(video_original_frames, matriz_homografia, (LARGURA_PERSPETIVA, ALTURA_PRESPETIVA))

reproduzir_frames(video_vista_topo, titulo="Perspetiva Vista de Cima")
```

<center><h2><b>Configuração das Vias e FPS</b></h2></center>

Uma vez obtida a vista de topo, a segmentação das vias torna-se muito simples, pois as linhas descontinuas divisórias agora estão verticais e equidistantes. Logo, dividimos a largura das imagens por 3 secções iguais para definir os limites das vias: 
- limite_v1 - limite entre a via da esquerda e a do meio
- limite_v2 - limite entre a via do meio e a que está mais à direita.

A função **obter_via** recebe a coordenada x onde o veículo se encontra no frame atual e devolve a via em que este se encontra (1 (esquerda), 2 (meio) ou 3(direita)).

```python
limite_v1 = int(LARGURA_PERSPETIVA / 3)      
limite_v2 = int(limite_v1 * 2) 

LIMITES_VIAS = [0, limite_v1, limite_v2, LARGURA_PERSPETIVA] 

#Func auxilar
def obter_via(x):
   #com base no x do veiculo recorna o indice da via em q está
    for i in range(len(LIMITES_VIAS) - 1):
        if LIMITES_VIAS[i] <= x < LIMITES_VIAS[i+1]:
            return i + 1
    return 0 #retorna 0 se for indefenido
```

Adicionalmente, para o cálculo da velocidade ($v = \Delta d / \Delta t$), é importante sabermos o intervalo de tempo exato entre frames com o numero real de frames por segundo. Utilizamos `cap.get(cv2.CAP_PROP_FPS)` para extrair o Frame Rate real do ficheiro de vídeo, o que até ao momento ainda não tinhamos realizado. Assim garantimos a precisão temporal. Se fossemos nós a decidir um frame rate, então as velocidades medidas não seriam as reais, já que bastaria aumentar o frame rate (acelerar o video) para as velocidades serem maiores

```python
#Verificamos qual os fps da imagem
cap = cv2.VideoCapture("AutoEstrada.avi" )

if cap.isOpened():
    fps = cap.get(cv2.CAP_PROP_FPS)
    print(f"FPS: {fps}")
else:
    print("Erro ao abrir o vídeo")

cap.release()
```

<center><h2><b>Algoritmo de Cálculo de Velocidade</b></h2></center>

A função **processar_velocidades** integra todo o conhecimento geométrico e físico adquirido. O processo para cada veículo em cada frame é o seguinte:

- Ponto de Contacto: Selecionamos o ponto $(x, y+h)$ da bounding box original. Este corresponde mais ou menos a um peneu (canto inferior esquerdo) que está realmente em contacto com a estrada, assim garante que o ponto está no plano Z=0 para a sua projeção homografica funcionar mais precisamente. O centro da caixa ou o topo não servem tão bem, pois a sua projeção não iria ser tão precisa porque estes pontos não estão em contacto com o chão.

- Transformação: Aplicamos **cv2.perspectiveTransform** apenas a este ponto de contacto, e obtemos as suas coordenadas $(x_t, y_t)$ na vista de topo.

- Identificação da Via: Com base na coordenada $x_t$, classificamos o veículo na via correspondente.

- Cálculo da Velocidade: Como calcular a velocidade instantânea entre frames consecutivos ($t$ e $t-1$) introduz ruído devido à pequena "vibração" da boundingbox então implementámos uma média de velocidade para saber a velocidadede "instantânea". Calculamos a distância percorrida entre o frame atual e o frame de há 5 instantes atrás (GAP_FRAMES = 5).A distância Euclidiana em píxeis (dist_pixeis) é convertida para metros (dividir pelos pixeis por metro previamente calculados) e a velocidade é obtida pela fórmula clássica $v = \frac{\Delta d}{\Delta t}$.

```python
#CALIBRAÇÃO 
LARGURA_VIA_PIXEIS = LARGURA_PERSPETIVA/3

#Pixeis por Metro
#Se uma via tem 3,75 metros então se fizermos regra de 3 simples sabemos que 1 metro = (larg da via / 3,75) pixeis
PPM = LARGURA_VIA_PIXEIS / 3.75

FPS = 23.97  # Frames por segundo do video original

def processar_velocidades(dados_rastreamento, matriz_homografia):
    historico_veiculos = {} 
    stats_vias = {1: [], 2: [], 3: []} 
    novos_dados_rastreamento = []

    for num_frame, veiculos_frame in enumerate(dados_rastreamento):
        novo_frame_dados = []
        
        for veiculo in veiculos_frame:
            # veiculo = (x, y, w, h, id_carro)
            x, y, w, h, id_carro = veiculo
            
            # Ponto de contacto - canto inferior esquerdo que corresponde a uma roda do carro
            ponto_base = np.array([[[x, y + h]]], dtype=np.float32) 
            #O perspectiveTransform recebe como parametros array 3d
            
            # Transformar ponto
            ponto_transf = cv2.perspectiveTransform(ponto_base, matriz_homografia)
            #O opencv está pronto para transformar muitos pontos de uma só vez, como só passamos um ponto ele está no primeiro indice
            x_t = ponto_transf[0][0][0]
            y_t = ponto_transf[0][0][1]
            
            #Histórico
            if id_carro not in historico_veiculos:
                historico_veiculos[id_carro] = []
            historico_veiculos[id_carro].append((num_frame, x_t, y_t))
            
            # Calcula
            velocidade_kmh = 0.0
            via = obter_via(x_t)

            #Se calcularmos a velocidade entre frames consecutivos, vai haver mais ruido por causa  da vibração da bounding box
            #Por isso decidimos, para calular a velocidade instantania, ver a distancia e o tempo que demorou a percorrer no espaço de 5 frames
            GAP_FRAMES = 5
            hist = historico_veiculos[id_carro]
            
            if len(hist) > GAP_FRAMES:
                frame_ant, x_ant, y_ant = hist[-1 - GAP_FRAMES] #o ultimo frame(-1) compara com 5 frames atrás
                
                #Calc da dist e intervalo tempo
                ponto_atual = (x_t, y_t)
                ponto_antigo = (x_ant, y_ant)
                
                dist_pixeis = math.dist(ponto_atual, ponto_antigo)#Dist que percurreu em pixeis
                dist_metros = dist_pixeis / PPM #Dist que percurreu em metros
                delta_tempo = (num_frame - frame_ant) / FPS #Tempo que passou em segundos
 
                distancia_km = dist_metros / 1000.0
                tempo_horas = delta_tempo / 3600.0 #3600 seg em 1hora
                velocidade_kmh = distancia_km / tempo_horas
                
                stats_vias[via].append(velocidade_kmh)

            nova_label = f"{id_carro}|{int(velocidade_kmh)}km/h"
            novo_frame_dados.append((x, y, w, h, nova_label))
            
        novos_dados_rastreamento.append(novo_frame_dados)

    # OUTPUT 
    print("VELOCIDADE MÉDIA EM CADA VIA:")
    for via, velocidades in stats_vias.items():
            media = sum(velocidades) / len(velocidades)
            print(f"Via {via}: Velocidade Média = {media:.2f} km/h")
        
    return novos_dados_rastreamento

# _______OUTPUT-TESTE___________
dados_com_velocidade = processar_velocidades(dados_com_veiculos_identificados, matriz_homografia)

reproduzir_frames(video_original_frames, pontos_roi, dados_com_velocidade, titulo="Resultado Final com Velocidades")
```

**Resultado Obtido:** Os resultados apresentam-se consistentes com o esperado para um cenário de autoestrada. As velocidades médias situam-se no intervalo típico (50 km/h a 120 km/h) valores limite de circulação dentro de uma autoestrada em Portugal. 

Observa-se que a via mais à esquerda (Via 1), tipicamente usada para realizar ultrapassagens, tem uma velocidade média superior às vias mais à direita, o que valida assim a lógica do sistema de contagem de velocidades implementado.
